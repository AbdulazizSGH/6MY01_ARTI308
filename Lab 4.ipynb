```python


import pandas as pd
import numpy as np 
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA



ds = pd.read_csv("ai_impact_jobs_2010_2025.csv")

ds.isna().sum()

#bunch of missing cells. Algortithm to use is unclear, so I'll just fill it with "None" for now.


ds["ai_keywords"] = ds["ai_keywords"].fillna("None")
ds["ai_skills"] = ds["ai_skills"].fillna("None")

ds.isna().sum()

#Using IQR to look for Outliers

plt.figure(figsize=(6,4))
sns.boxplot(x=ds['automation_risk_score'])
plt.title("automation_risk_score")
plt.show()


Q1 = ds['automation_risk_score'].quantile(0.25)
Q3 = ds['automation_risk_score'].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = ds[(ds['automation_risk_score'] < lower) | (ds['automation_risk_score'] > upper)]
outliers.head(15)

#This data set was cleaned for use in Machine learning so no outliers is expected

ds[["automation_risk_score","ai_intensity_score",]].head()


#Again, this dataset is cleaned. Normalization is already applied. I'll pretend it isn't and apply the techniques anyway

#from sklearn.preprocessing import MinMaxScaler

#scaler = MinMaxScaler()
#ds_scaled = ds[['automation_risk_score', 'ai_intensity_score']].copy()

#ds_scaled[['automation_risk_score', 'ai_intensity_score']] = scaler.fit_transform(df_scaled)

#ds_scaled.head()

#from sklearn.preprocessing import StandardScaler

#scaler = StandardScaler()
ds_standardized = ds[['automation_risk_score', 'ai_intensity_score']].copy()

#df_standardized[['automation_risk_score', 'ai_intensity_score']] = scaler.fit_transform(df_standardized)

#df_standardized.head()

plt.figure(figsize=(6,4))
sns.heatmap(ds_standardized[['automation_risk_score','ai_intensity_score']].corr(), 
            annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap (Before PCA)")
plt.show()

#From the figure we can see that automation risk score and intensity score aren't that related

#Using PCA



X = ds_standardized[['automation_risk_score', 'ai_intensity_score']]

pca = PCA(n_components=2)
principal_components = pca.fit_transform(X)

print("Explained Variance Ratio:", pca.explained_variance_ratio_)
```
